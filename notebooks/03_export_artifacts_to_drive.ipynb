# 03 — Export artifacts to Drive

Package run outputs and make sure the registry sheet is updated.

```python
import json
from pathlib import Path

RUN_ID = "RUN-YOLO-V0-001"
STREAM = "yolo"
RUN_DIR = (
    Path("/content/drive/MyDrive/AI Quantum HVTC/00_Data_Management/03_Training_Runs")
    / STREAM
    / RUN_ID
)
RUN_DIR.mkdir(parents=True, exist_ok=True)

manifest = {
    "run_id": RUN_ID,
    "stream": STREAM,
    "status": "placeholder",
    "notes": "Replace after a real run.",
}
(RUN_DIR / "run_manifest.json").write_text(
    json.dumps(manifest, ensure_ascii=False, indent=2), encoding="utf-8"
)
print(RUN_DIR)
```
